**Table 4 -- Evaluation des politiques apprises à chaque étape du curriculum learning**

```python
results = pd.DataFrame(
        index=[p.name for p in train_experiments],
        columns=[s.name for s in train_experiments],
        dtype=float,
    )

for policy in train_experiments:
    for scenario in train_experiments:

        exp = Experiment(
            name=policy.name,
            env_config=scenario.env_config,
            agent_config=scenario.agent_config,
            reward_config=scenario.reward_config,
            n_steps=scenario.n_steps,
        )

        history, _ = evaluate_sac(
            exp=exp,
            n_episodes=100,
            n_render=0,
        )

        df = pd.DataFrame(history)

        results.loc[policy.name, scenario.name] = df["success_rate"].mean()

results
```
